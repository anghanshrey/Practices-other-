#### **1. Reading and Writing CSV, JSON, and Excel Files**


##### Task 1: Read a CSV file into a DataFrame
##### File: 'sample_data.csv' (create your own with some dummy data if needed)

##### Task 2: Save the DataFrame into a new Excel file named 'output_data.xlsx'

##### Task 3: Load a JSON file named 'sample_data.json' into a DataFrame

##### Task 4: Convert a DataFrame into a JSON file named 'export_data.json'

```python
import pandas as pd
import numpy as np

data = pd.DataFrame({
    "Name": ["Alice", "Bob", "Charlie", "David"],
    "Age" : [21, 20, 22, 21],
    "City" : ["New York", "Paris", "London", "Sun Francisco"]
})

data.to_csv("sample_data.csv", index=False)

data1 = pd.read_csv("sample_data.csv")

data1.to_csv("output_data.xlsx", index=False)

data1.to_json("sample_data.json", index=False)

data_json = pd.read_json("sample_data.json")

data_json.to_json("export_data.json")
```

#### **2. Handling Missing Values, Duplicates, and Data Types**

##### Task 5: Load a DataFrame with missing values and fill them with appropriate values

##### Task 6: Drop rows where any value is missing

##### Task 7: Identify and remove duplicate rows

##### Task 8: Convert a column from string to numeric (handle errors gracefully)

##### Task 9: Check and change data types of columns in a DataFrame

```python
missing_value = pd.DataFrame({
    "Name": ["Alice", "Bob", "Alice", "David"],
    "Age" : ["21", "20", "21", np.nan],
    "City" : ["New York", "Pairs", "New York", np.nan],
    "salary" : ["20000", "25000", "20000", "70000"]
})

missing_value.to_csv("dirty_data.csv", index=False)

data = pd.read_csv("dirty_data.csv")

print(data)

data = data.dropna(subset=["Age"])

print(data)

data.duplicated().sum()

data.drop_duplicates(inplace=True)

print(data)

data["Age"] = pd.to_numeric(data["Age"], errors='coerce')

data['salary'] = pd.to_numeric(data["salary"], errors='coerce')

data.dtypes

data['Age'] = data['Age'].astype(int)

print(data)

data['salary'] = data['salary'].astype(int)

print(data)
```

#### **3. Handling Dates and Times using Pandas**

##### Task 10: Load a DataFrame containing a 'date' column with string format, and convert it to datetime

##### Task 11: Extract year, month, and day from the datetime column

##### Task 12: Filter the DataFrame for rows within a specific date range

##### Task 13: Create a new column with the difference in days between two datetime columns

```python
add_datetime= pd.DataFrame({
    "Name": ["Alice", "Bob", "Charlie", "David"],
    "Age" : [21, 20, 21, 22],
    "City" : ["New York", "Pairs", "New York", "Sun francisco"],
    "Salary" : [20000, 25000, 20000, 70000],
    "Date" : ["2025-01-01","2025-05-10","2025-06-15","2025-08-20"]
})

add_datetime.to_csv("datetime_data.csv", index=False)

data_dt = pd.read_csv("datetime_data.csv")

print(data_dt)
```

```python
data_dt['Date'] = pd.to_datetime(data_dt['Date'], errors='coerce')

data_dt.dtypes

data_dt['Date_year'] = data_dt['Date'].dt.year

print(data_dt)
```

```python
data_dt['Date_Month'] = data_dt['Date'].dt.month

print(data_dt)
```

```python
data_dt["Date_day"] = data_dt['Date'].dt.day

print(data_dt)
```
